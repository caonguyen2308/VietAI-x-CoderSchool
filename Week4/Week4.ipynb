```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Count the number of educational level 
df = pd.read_csv('Week4_1_data.csv')
grouped_df = df.groupby('educational_level')['customer_id'].count()
grouped_df
```

```python
# Sort 20 people have highest income
top_20_people = df.sort_values(by="annual_income", ascending=False).head(20)
top_20_people
```

```python
# Customers who were born after 1960 and have annual_income > 50,000
filtered_customers = df[(df['year_of_birth'] > 1960) & (df['annual_income'] > 50000)]
print(filtered_customers)
```

```python
top_20_people_2 = filtered_customers.sort_values(by='annual_income', ascending=False).head(20)
top_20_people
```

```python
average_income_by_education = df.groupby('educational_level')['annual_income'].mean()
average_income_by_education
```

As can be seen, the most people who get high income must be graduation from university. It doesn't mean that you have a high degree like master or PhD and you have a high income. Most people who graduation get higher income than the PhD and Master
