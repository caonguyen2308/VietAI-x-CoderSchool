```python
import pandas as pd
```

```python
df = pd.read_csv('dataset_2.csv')
df
```

```python
# Print the index of students whose math and reading scores are both in the Top 100
top_math_indices = df.nlargest(100, 'math score').index
top_reading_indices = df.nlargest(100, 'reading score').index

top_both = top_math_indices.intersection(top_reading_indices)

print("Index of students in Top 100 for both Math and Reading:")
print(top_both.tolist())
```

```python
# Print the indices of the top 20 students by writing score in each race/ethnicity group
top_20_writing_by_group = df.groupby('race/ethnicity').apply(lambda x: x.nlargest(20, 'writing score')).reset_index(drop=True)

print("Indices of top 20 students by writing score in each group:")
print(top_20_writing_by_group.index.tolist())
```

```python
# Count number of students in Group A by parental level of education
group_a = df[df['race/ethnicity'] == 'group A']
group_a_counts = group_a['parental level of education'].value_counts()

print("Number of Group A students by parental level of education:")
print(group_a_counts)
```

```python
# Is there a trend that higher parental education leads to higher average scores?
df['average score'] = df[['math score', 'reading score', 'writing score']].mean(axis=1)

# Group by parental education
avg_by_education = df.groupby('parental level of education')['average score'].mean().sort_values(ascending=False)

print("Average student scores by parental level of education:")
print(avg_by_education)
```

```python
# Does lunch quality affect average scores?
avg_by_lunch = df.groupby('lunch')['average score'].mean().sort_values(ascending=False)

print("Average student scores by lunch type:")
print(avg_by_lunch)
```

```python
# Find Top 10 students with highest math scores in each group
top_10_math_by_group = df.groupby('race/ethnicity').apply(lambda x: x.nlargest(10, 'math score')).reset_index(drop=True)

print("Indices of top 10 math scorers in each group:")
print(top_10_math_by_group.index.tolist())
```

```python
# For students with same group, parental education, and lunch, does test prep course help?
grouped = df.groupby(['race/ethnicity', 'parental level of education', 'lunch'])

results = []

for group_keys, group_data in grouped:
    prepared = group_data[group_data['test preparation course'] == 'completed']
    not_prepared = group_data[group_data['test preparation course'] == 'none']
    
    if len(prepared) > 0 and len(not_prepared) > 0:
        prepared_avg = prepared['average score'].mean()
        not_prepared_avg = not_prepared['average score'].mean()
        diff = prepared_avg - not_prepared_avg
        
        results.append({
            'group': group_keys,
            'avg score (completed)': prepared_avg,
            'avg score (none)': not_prepared_avg,
            'difference': diff
        })

result_df = pd.DataFrame(results)
print("Impact of test preparation course on average scores:")
print(result_df.sort_values(by='difference', ascending=False))
```
