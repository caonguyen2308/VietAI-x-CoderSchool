```python
import pandas as pd
```

```python
df = pd.read_csv('dataset_1.csv')
df
```

```python
# Sort 10 cities with highest population
top_10_pop = df.sort_values(by='Population', ascending=False).head(10)
print("Top 10 cities by population:")
print(top_10_pop[['City', 'Country', 'Population']])

# Top 10 cities with lowest population
bottom_10_pop = df.sort_values(by='Population', ascending=True).head(10)
print("\nBottom 10 cities by population:")
print(bottom_10_pop[['City', 'Country', 'Population']])
```

```python
# Country that have at least 3 cities
country_counts = df['Country'].value_counts()
countries_with_3_or_more = country_counts[country_counts >= 3]
print("\nCountries with at least 3 cities in the dataset:", countries_with_3_or_more)
```

```python
# Top 5 countries have the most cities in dataset
top_5_countries = country_counts.head(5)
print("\n🏆 Top 5 countries with the most cities in the dataset:")
print(top_5_countries)
```

```python
df['Population'] = df['Population'].replace(',', '', regex=True)
df['Population'] = pd.to_numeric(df['Population'], errors='coerce')
df['Area KM2'] = df['Area KM2'].astype(str).str.replace(',', '')
df['Area KM2'] = pd.to_numeric(df['Area KM2'], errors='coerce')
df = df.dropna(subset=['Population', 'Area KM2'])

# Get top 20 cities by population
top_20_pop = df.nlargest(20, 'Population')
top_20_area = df.nlargest(20, 'Area KM2')
common_cities = pd.merge(top_20_pop, top_20_area, on='City')

# Show the result
print("Cities ranked in Top 20 by both Population and Area:")
print(common_cities[['City', 'Country_x', 'Population_x', 'Area KM2_x']])
```

```python
# Average population density by country (per KM2)
df['Density KM2'] = df['Density KM2'].astype(str).str.replace(',', '')
df['Density KM2'] = pd.to_numeric(df['Density KM2'], errors='coerce')
density_by_country = df.groupby('Country')['Density KM2'].mean().sort_values(ascending=False)
print("\nAverage population density by country (per km2):")
print(density_by_country)
```

```python
# Largest city by population in countries with 2+ cities
multi_city_countries = df['Country'].value_counts()
multi_city_countries = multi_city_countries[multi_city_countries >= 2].index.tolist()

filtered_df = df[df['Country'].isin(multi_city_countries)]

largest_city_per_country = filtered_df.loc[filtered_df.groupby('Country')['Population'].idxmax()]
print("\n🌐 Largest city by population in each country (with 2+ cities):")
print(largest_city_per_country[['Country', 'City', 'Population']])
```
