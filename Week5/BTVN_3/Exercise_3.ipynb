```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('dataset_3.csv')
df
```

```python
# List products with revenue over 100,000 Euros
high_revenue = df[df['merchant_rating_count'] > 100_000]
print("Products with revenue > 100,000 EUR")
print(high_revenue[['title', 'merchant_rating_count']])
```

```python
# List products in the Top 100 units_sold AND in the top 10% highest rating
rating_threshold = df['rating'].quantile(0.90)

top_100_units = df.nlargest(100, 'units_sold')

top_products = top_100_units[top_100_units['rating'] >= rating_threshold]

print("Products in top 100 units_sold AND top 10% rating")
print(top_products[['title', 'units_sold', 'rating']])
```

```python
# Is there a correlation between number of ratings and units sold?
if 'rating_count' in df.columns:
    correlation = df['rating_count'].corr(df['units_sold'])
    print(f"Correlation between rating_count and units_sold: {correlation:.2f}")

    # Optional: visualize
    sns.scatterplot(data=df, x='rating_count', y='units_sold')
    plt.title('Rating Count vs Units Sold')
    plt.show()
else:
    print("'Rating_count' column not found in dataset.")
```

```python
# Top 50 best-selling products with rating between 3.0 and 4.0
mid_rating = df[(df['rating'] >= 3.0) & (df['rating'] <= 4.0)]
top_50_mid_rating = mid_rating.sort_values(by='units_sold', ascending=False).head(50)

print("Task 4: Top 50 best-selling products with rating between 3.0 and 4.0")
print(top_50_mid_rating[['title', 'rating', 'units_sold']])
```

```python
# Count how many products have a price over 50 Euros
count_over_50 = df[df['price'] > 50].shape[0]
print(f"Number of products with price over 50 EUR: {count_over_50}")
```
